```python
import random

N = 4

def random_state():
    return [random.randint(0, N - 1) for _ in range(N)]

def conflicts(state):
    count = 0
    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                count += 1
    return count

def hill_climbing():
    state = random_state()

    while True:
        current = conflicts(state)
        if current == 0:
            return state

        neighbors = []
        for col in range(N):
            for row in range(N):
                if state[col] != row:
                    new_state = state[:]
                    new_state[col] = row
                    neighbors.append(new_state)

        best = min(neighbors, key=conflicts)
        if conflicts(best) >= current:
            state = random_state()
        else:
            state = best

solution = hill_climbing()
print("Solution:", solution)
```
