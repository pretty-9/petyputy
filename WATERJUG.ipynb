```python
from collections import deque

def water_jug():
    visited = set()
    queue = deque([((0, 0), [])])

    while queue:
        (x, y), path = queue.popleft()

        if x == 2 or y == 2:
            path.append((x, y))
            print("Solution Path:")
            for step in path:
                print(step)
            return

        if (x, y) in visited:
            continue
        visited.add((x, y))

        next_states = [
            (4, y), (x, 3),
            (0, y), (x, 0),
            (max(0, x - (3 - y)), min(3, y + x)),
            (min(4, x + y), max(0, y - (4 - x)))
        ]

        for state in next_states:
            queue.append((state, path + [(x, y)]))

water_jug()
```
