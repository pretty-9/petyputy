```python
from itertools import permutations

graph = {
    'A': {'B': 10, 'C': 15, 'D': 20},
    'B': {'A': 10, 'C': 35, 'D': 25},
    'C': {'A': 15, 'B': 35, 'D': 30},
    'D': {'A': 20, 'B': 25, 'C': 30}
}

cities = list(graph.keys())
start = cities[0]

min_path = None
min_cost = float('inf')

for perm in permutations(cities[1:]):
    path = [start] + list(perm) + [start]
    cost = 0
    for i in range(len(path) - 1):
        cost += graph[path[i]][path[i + 1]]

    if cost < min_cost:
        min_cost = cost
        min_path = path

print("Best Path:", ' -> '.join(min_path))
print("Minimum Cost:", min_cost)
```
